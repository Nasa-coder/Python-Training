# If statements and decision-making

An if statement checks a condition and decides whether to run a block of code. Conditions evaluate to either True or False, and those booleans control the flow of the program.

```text
if score >= 70:
    print("Pass")
else:
    print("Retry")

Condition -> True  ?  run pass block
           False ?  run else block
```

### How to think about it
Every decision is a fork in the road. The computer checks one condition at a time, and only the matching branch runs. Try to read conditions as questions: “Is this true?”

# Making decisions with if statements

An **`if` statement** lets a program choose between different paths. Instead of running every line in order no matter what, Python can check a condition and only run a block when that condition is true. This is how programs respond to scores, passwords, prices, user roles, and almost every other real-world rule.

A condition is an expression that becomes either `True` or `False`. For example, `score >= 90` asks whether the value in `score` is at least 90. If it is true, the indented block under that `if` runs. If it is false, Python skips that block and continues.

You can add `elif` for additional checks and `else` for the fallback case. Python checks the branches **from top to bottom**, and the **first matching branch wins**. That is why more specific conditions usually come before broader ones.

When you read or write an `if` statement, ask two questions: **what rule am I testing, and what should happen when the rule is true?** That mindset makes conditionals much easier to design and debug.
